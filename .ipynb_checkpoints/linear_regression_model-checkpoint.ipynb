## Handling Categorical Features

### One Hot Encoding

```python
ohe = OneHotEncoder(drop = 'first',sparse_output=False,handle_unknown='ignore',dtype='int32')
```

```python
ohe_data = ohe.fit_transform(X_train[['sex','region']]) 
```

```python
ohe_data = pd.DataFrame(ohe_data,index=X_train.index)
```

```python
ohe_data
```

```python
X_train = pd.concat([X_train,ohe_data],axis = 1)
```

```python
X_train
```

### Ordinal Encoding

```python
oe = OrdinalEncoder(categories=[['yes','no']],dtype='int32')
```

```python
oe_data = oe.fit_transform(X_train[['smoker']])
```

```python
oe_data = pd.DataFrame(oe_data,columns=['smoking'],index = X_train.index)
```

```python
X_train = pd.concat([X_train,oe_data],axis = 1)
X_train
```

```python
X_train = X_train.drop(['sex','smoker','region'],axis=1)
```

```python
X_train.columns = X_train.columns.astype(str)
```

```python
X_train.columns
```

## Model Training

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```
